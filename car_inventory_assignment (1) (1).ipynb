```python
CARS_FILE = "CARS.txt"
def read_file():
    """Reads car records from the text file and returns a list of lists."""
    data = []
    try:
        with open(CARS_FILE, "r") as f:
            for line in f:
                line = line.strip()
                if line:
                    data.append(line.split(","))
    except FileNotFoundError:
        open(CARS_FILE, "w").close()
    return data


def write_file(data):
    """Writes the full list of cars back to the file."""
    with open(CARS_FILE, "w") as f:
        for record in data:
            f.write(",".join(record) + "\n")


def append_file(record):
    """Appends a single record to the file."""
    with open(CARS_FILE, "a") as f:
        f.write(",".join(record) + "\n")


# ---------------------- CAR MANAGEMENT FUNCTIONS ----------------------

def add_car():
    print("\n--- Add New Car ---")
    car_id = input("Enter Car ID: ")
    model = input("Enter Car Model: ")
    brand = input("Enter Brand: ")
    year = input("Enter Year: ")
    price = input("Enter Price per Day: ")

    append_file([car_id, model, brand, year, price])
    print("\nCar added successfully!\n")


def view_cars():
    print("\n--- All Cars ---")
    cars = read_file()

    if not cars:
        print("No cars available.\n")
        return

    for car in cars:
        print(f"\nID: {car[0]}\nModel: {car[1]}\nBrand: {car[2]}\nYear: {car[3]}\nPrice/Day: {car[4]}")
    print()


def search_car():
    print("\n--- Search Car ---")
    term = input("Enter Model or Brand to search: ").lower()

    cars = read_file()
    found = [car for car in cars if term in car[1].lower() or term in car[2].lower()]

    if not found:
        print("\nNo matching cars found.\n")
        return

    print("\n--- Results ---")
    for car in found:
        print(f"\nID: {car[0]}\nModel: {car[1]}\nBrand: {car[2]}\nYear: {car[3]}\nPrice/Day: {car[4]}")
    print()


def update_car():
    print("\n--- Update Car ---")
    car_id = input("Enter Car ID to update: ")

    cars = read_file()
    updated = False

    for car in cars:
        if car[0] == car_id:
            print(f"\nUpdating car: {car[1]}")

            model = input(f"New Model (current: {car[1]}): ")
            brand = input(f"New Brand (current: {car[2]}): ")
            year = input(f"New Year (current: {car[3]}): ")
            price = input(f"New Price/Day (current: {car[4]}): ")

            if model: car[1] = model
            if brand: car[2] = brand
            if year: car[3] = year
            if price: car[4] = price

            updated = True
            break

    if updated:
        write_file(cars)
        print("\nCar updated successfully!\n")
    else:
        print("\nCar ID not found.\n")


def delete_car():
    print("\n--- Delete Car ---")
    car_id = input("Enter Car ID to delete: ")

    cars = read_file()
    new_list = [car for car in cars if car[0] != car_id]

    if len(new_list) == len(cars):
        print("\nCar ID not found.\n")
        return

    write_file(new_list)
    print("\nCar deleted successfully!\n")


# ---------------------- MENUS ----------------------

def car_menu():
    while True:
        print("\n---- Car Management ----")
        print("1. Add Car")
        print("2. View Cars")
        print("3. Search Car")
        print("4. Update Car")
        print("5. Delete Car")
        print("6. Back to Main Menu")

        choice = input("Enter choice: ")

        if choice == "1": add_car()
        elif choice == "2": view_cars()
        elif choice == "3": search_car()
        elif choice == "4": update_car()
        elif choice == "5": delete_car()
        elif choice == "6": break
        else: print("Invalid input. Try again.\n")


def main_menu():
    while True:
        print("\n======== MAIN MENU ========")
        print("1. Car Management")
        print("2. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            car_menu()
        elif choice == "2":
            print("Exiting program... Goodbye!")
            break
        else:
            print("Invalid input.\n")


main_menu()
```
